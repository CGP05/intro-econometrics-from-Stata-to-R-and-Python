```python
import matplotlib.pyplot as plt
import numpy as np

#Set the number of observations that will be drawn to 10000 and the seed
# Roll a 6-sided die 10000 times
np.random.seed(3)
die = np.floor(np.random.uniform(1, 7, 10000)).astype(int)  # Corrected the upper limit to 7

# Generate the histogram for 10000 observations of your random variable
plt.bar(range(1, 7), np.bincount(die, minlength=7)[1:] / len(die) * 100, tick_label=range(1, 7), color='blue')
plt.title("Random Coin Flips (10000 Obs)")
plt.xlabel("Outcome")
plt.ylabel("Percentage")
plt.show()
```
