```python
#Define a variable z over the range 0 to 4 with 5 observations and all other values are missing (empty)

z = pandas.Series([0, 1, 2, 3, 4], index=range(5))
print(z)
```

```python
#List the values in a variable within a range
#list variable_name in (start)/(end)

print(list(enumerate(z, start=1)))
```

```python
#Create a variable k in Stata for the equation in your last tutorial k = (−((z − 2)^2) + 4)


k = -((z - 2) ** 2) + 4
print(k)
```

```python
##Calculate the definite integral over the observed z range using the trapezoidal rule

integral = numpy.trapezoid(k.to_numpy(), x=z.to_numpy())
print(integral)
```

```python
##Generate kprime as the numerical derivative of k with respect to z

kprime = pandas.Series(numpy.gradient(k.to_numpy(), z.to_numpy()), index=z.index)
print(kprime)
```
